```python
import random

otp = random.randint(100000, 999999)

print("Your OTP is:", otp)

for attempt in range(3):

    user_otp = int(input("Enter OTP: "))

    if user_otp == otp:
        print("✅ OTP Verified Successfully!")
        break
    else:
        remaining = 2 - attempt
        print("❌ Wrong OTP!")

        if remaining > 0:
            print("Attempts remaining:", remaining)
        else:
            print("🚫 Maximum attempts reached.")
```

###### Summary

This project is a simple OTP Verification System developed using Python. The random library is used to generate a random 6-digit OTP. The user is given a maximum of 3 attempts to enter the correct OTP. The program compares the entered OTP with the generated OTP and displays whether the OTP is verified successfully or incorrect. If all three attempts are used incorrectly, the program shows a maximum attempts message.

This project helps in understanding important Python concepts such as random number generation, input handling, conditional statements, loops, variables, and the break statement.
